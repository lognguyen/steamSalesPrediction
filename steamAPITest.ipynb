```python
def release_date(x):
    now = dt.date.today()
    date = ast.literal_eval(x)['date'][-4:]
    if (now.year - int(date) < 20):
        return 1
    else: 
        return 0
```

```python
def categories_function(categories, attr):
    try:
        description = ''
        temp_array = ast.literal_eval(str(categories))
        for object_ar in temp_array:
            description += ", " + object_ar[attr]
        return (description)
    except SyntaxError:
        print("WARNING CATE")
        return str(attr)
```

```python
def emptyString(support_info):
    try: 
        car = ast.literal_eval(str(support_info))
        if not car['email'] and not car['url']:
            return 0
        return 1
    except SyntaxError:
        print("WARNING support_info")
        return 1
```

```python
def developers_join(developers):
    try: 
        return ', '.join(ast.literal_eval(str(developers)))
    except:
        print("WARNING DEV")
        return str(developers)
```

```python
def number_clean(column_to_apply,name):
    try:
        return int(ast.literal_eval(column_to_apply)[name])
    except:
        print("WARNING recommendation or price")
        return 0
```

```python
def content_warning_clean(content_descriptors):
    try:
        notes = ast.literal_eval(str(content_descriptors))['notes']
        if not notes:
            return 0
        return 1
    except:
        print("WARNING content_descriptors")
        return 0
```

```python
def achivemet_clean(achivemet_clean):
    try:
        total = ast.literal_eval(str(achivemet_clean))['total']
        if not total:
            return 0
        return total
    except:
        print("WARNING achivemet_clean")
        return 0
```

```python
def platform(plat):
    temp_str = ''
    new_plat = ast.literal_eval(str(plat))
    wind = new_plat['windows']
    mac = new_plat['mac']
    linux = new_plat['linux']
    if wind == True:
        temp_str += 'windows,'
    if mac == True:
        temp_str += 'mac,'
    if linux == True:
        temp_str += 'linux,'
    return temp_str.rstrip(',')
```

```python
def bagFunction(column_name):
    bag_of_words = []
    for row in result_df[column_name].values:
        if (row != None):
            temp = row.strip().split(", ")
            bag_of_words = bag_of_words + temp
#         bag_of_words = list(filter(None, bag_of_words))
    return bag_of_words
```

```python
# lang = bagFunction('supported_languages')
bag_of_languages = Counter(bagFunction('supported_languages')).most_common(3)
```

```python
# result_df['supported_languages'].head()
# result_df['supported_languages'].values.flatten()
german = bag_of_languages[1][0]
french = bag_of_languages[2][0]
english = bag_of_languages[0][0]
column_english = result_df['supported_languages'].apply(lambda x: 1 if english in x else 0)
column_english.name  = 'English'
column_german = result_df['supported_languages'].apply(lambda x: 1 if german in x else 0)
column_french = result_df['supported_languages'].apply(lambda x: 1 if french in x else 0)
column_german.name  = 'German'
column_french.name  = 'French'
```

```python
result_df = pd.concat([result_df,column_french,column_english,column_german],axis=1, join='inner')
result_df = result_df.drop(columns = ["supported_languages"])
result_df.head()
```
